Step 2: Generate and Save 10,000 Simulated Ad Records

```python
import sys
sys.path.append("../src")

from data_generator import generate_ad_data, save_raw_data

# Generate 10,000 simulated records
df = generate_ad_data(10_000)
print(f"Shape: {df.shape}\n")
print(df.head(5).to_string(index=False))

# Validation checks
checks = {
    "No missing values": df.isna().sum().sum() == 0,
    "Clicks <= impressions": (df["clicks"] <= df["impressions"]).all(),
    "Conversions <= clicks": (df["conversions"] <= df["clicks"]).all(),
    "Spend > 0": (df["spend"] > 0).all(),
    "Revenue >= 0": (df["revenue"] >= 0).all(),
}
print("\nValidation checks")
for name, passed in checks.items():
    print(f"  {name:<24}{'PASS' if passed else 'FAIL'}")

# Save (refuses to overwrite an existing file)
try:
    save_raw_data(df, "../data/raw/ads_10k.csv", "../data/backup")
    print("\nSaved: data/raw/ads_10k.csv (backup copy in data/backup)")
except FileExistsError as error:
    print(f"\n{error}")
```

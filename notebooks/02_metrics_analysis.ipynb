Step 3: Calculate Advertising Metrics by Platform (10,000 Records)

```python
import sys
sys.path.append("../src")

import pandas as pd
from metrics import summarize

# Load raw data (read-only)
df = pd.read_csv("../data/raw/ads_10k.csv", parse_dates=["date"])

# Metrics per platform, best ROAS first
platform_summary = summarize(df, "platform").sort_values("roas", ascending=False)

# Readable English labels for display and saving
labels = {
    "spend": "Spend (Rs.)",
    "revenue": "Revenue (Rs.)",
    "profit": "Profit (Rs.)",
    "ctr_pct": "CTR (%)",
    "cpc": "CPC (Rs.)",
    "cpa": "CPA (Rs.)",
    "cvr_pct": "Conversion Rate (%)",
    "roas": "ROAS",
}
table = platform_summary[list(labels)].rename(columns=labels)

print("Platform performance (10,000 records)\n")
print(table.to_string(float_format=lambda x: f"{x:,.2f}"))

# Save the table (safe to regenerate; it is not raw data)
table.to_csv("../outputs/tables/platform_metrics_10k.csv")
print("\nSaved: outputs/tables/platform_metrics_10k.csv")
```
